```python
import seaborn as sns
```

```python
#bascic plotting with seaborn
tips = sns.load_dataset('tips')
tips
```

```python
# Create a scatter plot
import matplotlib.pyplot as plt
sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title('scatter plot of total_bill vs tip ')
```

```python
# Line plot 
sns.lineplot(x='size',y='total_bill',data=tips)
plt.title('line plot of size vs total_bill')
#plt.show()
```

```python
# Catagrical plots
# Bar Plot
sns.barplot(x='day',y='total_bill',data=tips)
plt.title('Bar Plot of Day v/s total_bill')
```

```python
sns.boxplot(x='day',y='total_bill',data=tips)
plt.title('Box Plot of day v/s total_bill')
```

```python
#violin plot

sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
# histogram plot
sns.histplot(tips['total_bill'],bins=20,kde=True)
```

```python
# kernel density plot or KDE plot
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
# Pair Plot
sns.pairplot(tips)
```

```python
# Heatmap
corr=tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```
